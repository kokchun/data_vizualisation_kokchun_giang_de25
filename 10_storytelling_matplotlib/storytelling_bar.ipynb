# Data storytelling bar chart

- read in the dataset
- calculate total sold cars per brand
- return a df from it

```python
import pandas as pd
from constants import GRAY_1, GRAY_2, GRAY_3, BLUE_1
import duckdb

df = pd.read_csv("norway_new_car_sales_by_make.csv")

df.head()
```

## Pick out total quantity per brand

```python
df_qty_brand = duckdb.sql("""--sql
    SELECT 
        make AS Brand,
        SUM(quantity) AS Quantity
    FROM df
    GROUP BY brand
    ORDER BY quantity DESC
"""
).df().set_index("Brand")

df_qty_brand.head()
```
